```python
import numpy as np
from dose_data import dose
from mask_data import mask

import matplotlib.pyplot as plt

def plot_dvh(dose, mask, bins=100, ax=None):
    """Rita en kumulativ DVH för dosvärden inom ett maskerat område."""
    dose = np.asarray(dose)
    mask = np.asarray(mask, dtype=bool)

    if dose.shape != mask.shape:
        raise ValueError("dose och mask måste ha samma form.")

    values = dose[mask]
    values = values[np.isfinite(values)]

    if values.size == 0:
        raise ValueError("Masken innehåller inga giltiga dosvärden.")

    thresholds = np.linspace(values.min(), values.max(), bins + 1)
    volume_percent = np.array([
        100 * np.mean(values >= threshold)
        for threshold in thresholds
    ])

    if ax is None:
        _, ax = plt.subplots()

    ax.step(thresholds, volume_percent, where="post")
    ax.set_xlabel("Dos")
    ax.set_ylabel("Volym över dosnivån (%)")
    ax.set_title("Kumulativ DVH")
    ax.set_ylim(0, 100)
    ax.grid(True)

    return ax
```
